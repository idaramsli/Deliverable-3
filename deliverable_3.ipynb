Task 1.

```python
from character import Character
""" 
Testing the character class, with creating characters, attacking, 
taking damage until HP reaches 0 then checking if the character is alive, 
and healing without going above hp_max.     
"""


ola = Character("Ola", hp_max=50, damage=8)
kari = Character("Kari", hp_max=40, damage=6)

print(ola)
print(kari)
print("Ola is alive:", ola.is_alive)

print("\nOla attacks Kari:")
ola.attack(kari)
print(kari)

print("\nKari takes on 40 damage:")
kari.take_damage(40)
print(kari)
print("Kari is alive:", kari.is_alive)

print("\nOla heals 20:")
ola.take_damage(15)
print("After damage:", ola)
ola.heal(20)
print("After heal:", ola)
```

Discussion:

In this task we updated the Character class so it follows better object-oriented principles. 
Instead of letting attack() change another character’s private _hp directly, we made a 
take_damage() method that only changes the character’s own HP. We also added heal() and 
the read-only property is_alive. 

This makes the code more encapsulated and safer, because no object can change another 
character’s HP directly. We also made sure that HP never goes below 0 or above hp_max 
by using max() and min().

Task 2

```python
from dice import Dice
""" 
Testing the Dice class, with using the __repr__ method, the .roll() method, and the __call__ method. 
Then showing an example of rolling a d4 ten times. 
Finally, testing the error handling by trying to create a dice with 1 side. 
"""

d6 = Dice()                 
d20 = Dice(sides=20)
d4 = Dice(4)

print(d6)                   
print(d20)                  
print(d4)                   


print("\nRolling d6 five times with .roll():")
print([d6.roll() for _ in range(5)])


print("\nCalling d20() five times (uses __call__):")
print([d20() for _ in range(5)])


print("\nRolling d4 ten times:")
print([d4.roll() for _ in range(10)])

try:
    Dice(sides=1)
except ValueError as error:
    print(f"Error: {error}")
```

Discussion: 
In this task we created a Dice class that can make dice with different number of sides 
(minimum 2). We implemented the roll() method to get random numbers, and also added __call__ so we can write d20() instead of d20.roll(). We used __repr__ so the dice objects print in a clear way, for example Dice(sides=6). 
To avoid invalid dice we raise a ValueError if someone tries to create a dice with 
fewer than 2 sides.
