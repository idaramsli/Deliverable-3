```python
from dice import Dice
""" Testing the Dice class"""

d6 = Dice()                 
d20 = Dice(sides=20)
d4 = Dice(4)

# Using _repr_ to print the dice objects
print(d6)                   
print(d20)                  
print(d4)                   

# Rolling the dice using the .roll() method
print("\nRolling d6 five times with .roll():")
print([d6.roll() for _ in range(5)])

# Rolling the dice by calling the object directly (__call__)
print("\nCalling d20() five times (uses __call__):")
print([d20() for _ in range(5)])

# example
print("\nRolling d4 ten times:")
print([d4.roll() for _ in range(10)])



# testing error handling
try:
    Dice(sides=1)
except ValueError as error:
    print(f"Error: {error}")
```

Discussion: 

In this task we created three differetn dices, with different number of sides. When printing them we made them show up woth the number of sides, to make it more clear which dice we are working with. 

The amount of sides is different, and therefore the result will be very different as well. The numbers are randim every time we run the code. To make sure that it didnt print dices with only one side we made a safety net so that it prints false, if n<2.

```python
from character import Character, Warrior, Mage
from dice import Dice
from items import Weapon

"""Warrior class"""

# Create a Warrior
warrior = Warrior("Warrior")

# Display the Warrior's attributes
print("Name:", warrior.name)
print("HP:", warrior.hp)
print("Armor:", warrior.armor)

# Test the armor by taking 10 damage
warrior.take_damage(10)
print()
print("HP after taking 10 damage:", warrior.hp)
print("------------------------------")

"""Mage class"""

mage = Mage("Mage")

# Display the Mage's attributes
print("Name:", mage.name)
print("HP:", mage.hp)   
print("Mana:", mage.mana)
print("Maximum Mana:", mage.mana_max)

# Create an enemy character to attack
enemy = Character("Enemy")

# Perform a magical attack on the enemy
enemy_hp_before = enemy.hp
mage.attack(enemy)

print()
print("Damage dealt:", enemy_hp_before - enemy.hp)
print("Enemy HP after magical attack:", enemy.hp)
print("Mage mana after magical attack:", mage.mana)


# Equip the Mage with a weapon
staff = Weapon("Magic Staff", Dice(6))
mage.weapon = staff

# Reduce mana below 10
mage.mana = 5

# Perform a normal attack
enemy_hp_before = enemy.hp
mage.attack(enemy)

print()
print("Damage dealt:", enemy_hp_before - enemy.hp)
print("Enemy HP after normal attack:", enemy.hp)
print("Mage mana after normal attack:", mage.mana)

```
